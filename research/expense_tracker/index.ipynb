```python
import os

# File to store expenses
FILE_NAME = "expenses.txt"

# Function to load expenses from file
def load_expenses():
    expenses = []
    if os.path.exists(FILE_NAME):
        try:
            with open(FILE_NAME, "r") as file:
                for line in file:
                    date, amount, category, description = line.strip().split("|")
                    expenses.append({
                        "date": date,
                        "amount": float(amount),
                        "category": category,
                        "description": description
                    })
        except Exception as e:
            print("Error loading expenses:", e)
    return expenses

# Function to save expenses to file
def save_expenses(expenses):
    try:
        with open(FILE_NAME, "w") as file:
            for exp in expenses:
                line = f"{exp['date']}|{exp['amount']}|{exp['category']}|{exp['description']}\n"
                file.write(line)
        print("Expenses saved successfully.")
    except Exception as e:
        print("Error saving expenses:", e)

# Function to add a new expense
def add_expense(expenses):
    try:
        date = input("Enter date (YYYY-MM-DD): ")
        amount = float(input("Enter amount: "))
        category = input("Enter category (Food, Travel, Bills, etc.): ")
        description = input("Enter description: ")
        expense = {
            "date": date,
            "amount": amount,
            "category": category,
            "description": description
        }
        expenses.append(expense)
        print("Expense added successfully!\n")
    except ValueError:
        print("Invalid input! Please enter the correct data type.")

# Function to view all expenses
def view_expenses(expenses):
    if not expenses:
        print("No expenses found.")
    else:
        print("\n----- All Expenses -----")
        for i, exp in enumerate(expenses, 1):
            print(f"{i}. {exp['date']} | {exp['amount']} | {exp['category']} | {exp['description']}")
        print()

# Main menu
def main():
    expenses = load_expenses()
    while True:
        print("\n--- Personal Expense Tracker ---")
        print("1. Add Expense")
        print("2. View Expenses")
        print("3. Save and Exit")

        choice = input("Choose an option (1/2/3): ")

        if choice == "1":
            add_expense(expenses)
        elif choice == "2":
            view_expenses(expenses)
        elif choice == "3":
            save_expenses(expenses)
            break
        else:
            print("Invalid choice. Try again.")

if __name__ == "__main__":
    main()
```
