```python
price = data['Mrp'].to_numpy()
print(price)
```

```python
print(np.mean(price))
print(np.median(price))
print(np.std(price))
```

```python
plt.scatter(data['Mrp'], data['Sale Price'], alpha=0.7)
plt.xlabel('Mrp')
plt.ylabel('Sale Price')
plt.title('Histogram of Mrp')
plt.show()
```

```python
plt.hist(data['Mrp'], bins=10, alpha=0.5, label='Mrp')
plt.title("Distribution of Apple Product Mrp")
plt.xlabel("Mrp")
plt.ylabel("Count")
plt.show()
```

```python
plt.bar( data['Product Name'].head(5), price[:5], color='green')
plt.title('Apple product phone price range')
plt.xlabel('Product Name')
plt.ylabel('Price')
plt.show()
```
