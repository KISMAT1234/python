```python
class Football:   
    def set_player(self, playerName):
        self.playerName = playerName
        # print(playerName)
    def get_player(self):
        print(self.playerName)


data = Football()
data.set_player("Lionel Messi")
data.get_player()
```

```python
class Rectangle:
    def calculate(self, l, b):
        return 2 * l + b

data = Rectangle()
value = data.calculate(5, 6)
print(value)
```

```python
class Dog:
    # creating constructor in python should use __init__
    def __init__(self, name, age):  
        print(self, "self printed")
        self.name = name
        self.age = age
        print(name, age)
        
    def bark(self):
        print("Woof!", self.name)
    
dog1 = Dog("Fido", 3)
dog1.bark()
```
