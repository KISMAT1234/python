```python
class Dog:
    # creating constructor in python should use __init__
    def __init__(self, name, age):  
        print(self, "self printed")
        self.name = name
        self.age = age
        print(name, age)
        
    def bark(self):
        print("Woof!", self.name)
    
dog1 = Dog("Fido", 3)
dog1.bark()
```
