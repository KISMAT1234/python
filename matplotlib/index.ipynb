```python
data = {
    "Height": [150,160,170,180,190,175,165,155],
    "Weight":[50,60,65,80,90,75,62,55],
    "Age":[80,650,60]
}

df = pd.DataFrame(data)
print(df,'convert to data frameeee')
plt.figure(figsize=(8,6))
plt.scatter(df['Height'], df['Weight'], df['Age'], color= 'blue')
plt.show()
```

```python
data = {
    'Income':[100,200,300,600],
    'Age':[1,2,3,9]
}
df = pd.DataFrame(data)

Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)

IQR = Q3 - Q1
print(Q1, 'q1')
print(Q3, '3')
print(IQR, 'iqr')

outliers = ((df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR)))
print(outliers)
```

```python
import matplotlib.pyplot as plt

# Data
streams = ['Science', 'Commerce', 'Arts', 'Computer Science', 'Others']
students = [120, 80, 100, 60, 40]

# Explode the 'Science' slice
explode = [0.1, 0, 0, 0, 0]  # Explode first slice (Science)

# Create pie chart
plt.figure(figsize=(8, 8))  # Make it square to ensure circle
plt.pie(
    students,
    labels=streams,
    autopct='%1.1f%%',  # Show percentages with 1 decimal
    startangle=90,      # Start from 90 degree angle
    explode=explode,    # Explode configuration
    shadow=True         # Optional: add shadow for 3D effect
)

plt.title('Percentage Distribution of Students in Academic Streams')
plt.axis('equal')  # Ensures the pie is a circle
plt.show()
```
