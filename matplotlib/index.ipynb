```python
import matplotlib.pyplot as plt
x = [3,5,2,6,7,1,2,7,1,7]
y = [80,90,75,80,90,50,65,85,40,100]
plt.scatter(x,y,color="red", marker='1')
plt.show()
```

```python
x=["A","B","C","D"]
y=[3,8,1,100]
plt.bar(x,y,color="red")
plt.show()
```

```python
data =[1,2,2,3,3,3,4,4,5,10]
plt.hist(data, bins=5, color="green",density=True, edgecolor="black")
plt.title("Histogram")
plt.xlabel("values")
plt.ylabel("Frequency")
plt.show()
```
