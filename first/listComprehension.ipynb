```python
numbers = [1, 2, 3, 4, 5]
doub = [num * 2 for num in numbers]
add = [data + 2 for data in numbers]
cube = [data ** 3 if data % 2 == 0 else data + 10 for data in numbers ]
print(doub)
print(add)
print(cube)
```

```python
def cube(value):
    result = []
    for i in value:
       if i % 2 == 0:
        result.append(i**3)
       else:
         result.append(i) 
    return result

variable = cube([1,2,3,4,5,6,7,8,9])
print(variable)
```

```python
word = "Python is a programming language"
print(list(word))
vowels = ['a','e','i','o','u']

def checkVowel(word):
     answer = []
     for i in word:
        if i in vowels:
            answer.append(i)
     return answer
data = checkVowel(word)
print(data)
```
