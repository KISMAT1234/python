```python
import re
s = "abc 456 @#$"
match = re.search(r'\d+', s)
print(match)
if match:
    num = int(match.group())
    print(num) 
else: 
    print("No number found")
```

```python
user = input("Enter your name: ")
vowel = ['a','e','i','o','u']
consonant = ['b','c','d','f','g','h','j','k','l','m','n','p','q','r','s','t','v','w','x','y','z']

if user in vowel:
    print("the word is vowel")
elif user in consonant:
    print("the word is consonant")
else:
    print("the word is neither vowel nor consonant")
```
