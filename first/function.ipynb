```python
def check(num):
    if(num % 2 == 0):
       return True
    else:
        return False
value = 10
data = check(value)
if(data):
    print("Number is even")
else:
    print("Number is odd")
```

```python
value1 = int(input("Enter a number: "))
value2 = int(input("Enter another number: "))
def add(a,b):
    return a+b
def sub(a,b):
    return a-b
def mul(a,b):
    return a*b
def div(a,b):
    return a/b

print(add(value1,value2))
print(sub(value1,value2))
print(mul(value1,value2)) 
print(div(value1,value2))

```

```python
# variable-length arguments
def add(*numbers):
      print(numbers)
      total = sum(numbers)
      print(total)
add(1,2,3,4,5)
```
