```python
day1 = ["Alice","Bob", "Charlie", "David", "Eva"]
day2 = ["Charlie", "Eva", "Frank", "Grace", "Bob"] 

s1 = set(day1)
s2 = set(day2)
print(s1)
print(s2)

both = s1 & s2
print(both)

only = s1 - s2
print(only)
```
