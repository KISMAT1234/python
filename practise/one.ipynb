```python
a = input("enter the letter")
letter = ['a','e','i','o','u']

countV = 0
countC = 0
for i in list(a):
    print(i)
    if i in letter:
        countV += 1
    else:
        countC += 1
    
print("Total vowels",countV)
print("Total consonants",countC)
```
