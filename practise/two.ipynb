```python
def is_even(num):
    return num % 2 == 0

if is_even(5):
    print("Even number")
else:
    print("Odd number")
```

```python
students = [
    ['Harry', 37.21],
    ['Berry', 37.21],
    ['Tina', 37.2],
    ['Akriti', 41],
    ['Harsh', 39]
]

all_grades = []
for student in students:
    grade = student[1]
    if grade not in all_grades:
        all_grades.append(grade)

all_grades.sort()
print(all_grades)

lowest = all_grades[1]
print(lowest)

low_student = []
for student in students:
    name = student[0]
    grade = student[1]
    if grade == lowest:
        low_student.append(name)

low_student.sort()
print(low_student[0], "first")
print(low_student[1], "second")


```
