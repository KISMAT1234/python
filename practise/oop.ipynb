```python
grade_levels = [
        (90, "You are a good student"),
        (80, "You are a medium student"),
        (70, "You are a bad student"),
        (60, "You are a poor student"),
        (50, "You are a dumb student"),
        (0, "You are a fail student") 
]
class Result:
    # disct  = {}
   ok = []
   def __init__(self, value):
        self.disct  = {}
        for i in range(value):
            subject = input("Enter the subject name:")
            subMarks = int(input("Enter the marks"))
            self.disct.update({
                subject: subMarks,
            })
        
   def calculateMarks(self):
        for key, value in self.disct.items():
            print(f"Your marks for {key} is {value}")

        total = sum(self.disct.values())
        count = len(self.disct)
        print("Your total marks is ",total)
        self.avg = total / count
        print("Your total average marks is",self.avg)

   def printGrade(self):
       for grade , messages in grade_levels:
           if self.avg >= grade:
               print(messages)
   

value = int(input("Enter number of subject you have:"))
re = Result(value)
re.calculateMarks()
re.printGrade()
    

```

```python
def get_student_grade(avg):
    grade_levels = [
        (90, "You are a good student"),
        (80, "You are a medium student"),
        (70, "You are a bad student"),
        (60, "You are a poor student"),
        (50, "You are a dumb student"),
        (0, "You are a fail student") 
    ]

    for firstData, messages in grade_levels:
        print(firstData, messages)
        if avg >= firstData:
            return messages

# Example usage
avg = 73
print(get_student_grade(avg))
```
